```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# Ensure Plotly works in VS Code Interactive Window
pio.renderers.default = "notebook_connected"  # or "iframe_connected"

# Create the DataFrame
data = {
    "species": ["Gentoo", "Gentoo", "Gentoo", "Adelie", "Gentoo"],
    "island": ["Biscoe", "Biscoe", "Biscoe", "Biscoe", "Biscoe"],
    "bill_length_mm": [43.4, 50.4, 46.4, 37.7, 50.7],
    "bill_depth_mm": [14.4, 15.7, 15.6, 16.0, 15.0],
    "flipper_length_mm": [218.0, 222.0, 221.0, 183.0, 223.0],
    "body_mass_g": [4600.0, 5750.0, 5000.0, 3075.0, 5550.0],
    "sex": ["Female", "Male", "Male", "Female", "Male"]
}

df = pd.DataFrame(data)

# -----------------------------
# 1️⃣ Histogram of Bill Length
# -----------------------------
fig_hist = px.histogram(
    df,
    x="bill_length_mm",
    nbins=5,
    color="species",
    title="Histogram of Bill Length",
    labels={"bill_length_mm": "Bill Length (mm)"}
)
fig_hist.show()

# -----------------------------
# 2️⃣ Bar Chart of Body Mass by Species
# -----------------------------
fig_bar = px.bar(
    df,
    x="species",
    y="body_mass_g",
    color="sex",
    barmode="group",
    title="Body Mass by Species and Sex",
    labels={"body_mass_g": "Body Mass (g)"}
)
fig_bar.show()

# -----------------------------
# 3️⃣ Pie Chart of Species Distribution
# -----------------------------
fig_pie = px.pie(
    df,
    names="species",
    color="species",
    title="Species Distribution",
    hole=0.3  # donut chart style
)
fig_pie.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create the DataFrame
data = {
    "species": ["Gentoo", "Gentoo", "Gentoo", "Adelie", "Gentoo"],
    "island": ["Biscoe", "Biscoe", "Biscoe", "Biscoe", "Biscoe"],
    "bill_length_mm": [43.4, 50.4, 46.4, 37.7, 50.7],
    "bill_depth_mm": [14.4, 15.7, 15.6, 16.0, 15.0],
    "flipper_length_mm": [218.0, 222.0, 221.0, 183.0, 223.0],
    "body_mass_g": [4600.0, 5750.0, 5000.0, 3075.0, 5550.0],
    "sex": ["Female", "Male", "Male", "Female", "Male"]
}

df = pd.DataFrame(data)

# Set Seaborn style
sns.set(style="whitegrid")

# Create scatter plot
plt.figure(figsize=(8, 6))
scatter = sns.scatterplot(
    data=df,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    size="body_mass_g",
    sizes=(50, 300),  # min/max point size
    palette="deep",
    edgecolor="black"
)

# Add title
plt.title("Penguin Bill Length vs Flipper Length (Sized by Body Mass)", fontsize=14)
plt.legend(bbox_to_anchor=(1.05, 1), loc=2)  # Move legend outside
plt.tight_layout()
plt.show()
```
